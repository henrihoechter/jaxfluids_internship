## newton solver

```python

T_tr_initial = 3000.0  # K - initial translational-rotational temperature
T_ve_initial = 10000.0   # K - initial vibrational-electronic temperature
dof_tr = 5              # fixed: 3 translational + 2 rotational for N2

species_bird_with_el = chemistry_utils.load_species_table(
    general_data_path=general_species_data_path,
    species_names=species_names,
    energy_model_config=energy_models.EnergyModelConfig(
        model="bird",
        include_electronic=True,
        data_path=bird_equilibrium_enthalpy_data_path,
    ),
)
e_ve_bird_with_el = species_bird_with_el.energy_model.e_ve(T)[0]
dof_ve_bird_with_el = calculate_dof(e_ve_bird_with_el, T, species.molar_masses[0])

cases = {
    "Bird (no electronic)":   dof_vib_bird,
    "Bird (with electronic)": dof_ve_bird_with_el,
    "Gnoffo (vib + el)":      dof_vib_gnoffo,
}

# ---- Newton solver ----
def newton_equilibrium_T(T_arr, dof_ve_arr, dof_tr, T_tr_init, T_ve_init, n_iter=50, tol=1e-8):
    """Solve: (dof_tr + dof_ve(T_eq)) * T_eq = dof_tr * T_tr_init + dof_ve(T_ve_init) * T_ve_init"""
    dof_ve_init = float(jnp.interp(T_ve_init, T_arr, dof_ve_arr))
    E_initial = dof_tr * T_tr_init + dof_ve_init * T_ve_init

    # Initial guess: treat dof_ve as constant at its initial value
    T_eq = E_initial / (dof_tr + dof_ve_init)

    for i in range(n_iter):
        dof_ve = float(jnp.interp(T_eq, T_arr, dof_ve_arr))
        f = (dof_tr + dof_ve) * T_eq - E_initial

        # Central-difference derivative of f w.r.t. T_eq
        dT = 1.0  # K
        dof_ve_p = float(jnp.interp(T_eq + dT, T_arr, dof_ve_arr))
        dof_ve_m = float(jnp.interp(T_eq - dT, T_arr, dof_ve_arr))
        df = ((dof_tr + dof_ve_p) * (T_eq + dT) - (dof_tr + dof_ve_m) * (T_eq - dT)) / (2 * dT)

        T_new = T_eq - f / df
        if abs(T_new - T_eq) < tol:
            return T_new, i + 1
        T_eq = T_new

    return T_eq, n_iter

print(f"Initial conditions: T_tr = {T_tr_initial} K,  T_ve = {T_ve_initial} K\n")
for name, dof_ve_arr in cases.items():
    T_eq, iters = newton_equilibrium_T(T, dof_ve_arr, dof_tr, T_tr_initial, T_ve_initial)
    dof_ve_init = float(jnp.interp(T_ve_initial, T, dof_ve_arr))
    dof_ve_eq   = float(jnp.interp(T_eq,         T, dof_ve_arr))
    dof_tot_init = dof_tr + dof_ve_init
    dof_tot_eq   = dof_tr + dof_ve_eq
    T_overall_init = (dof_tr * T_tr_initial + dof_ve_init * T_ve_initial) / dof_tot_init
    print(f"{name}:")
    print(f"  T_overall_initial = {T_overall_init:.2f} K")
    print(f"  T_eq              = {T_eq:.2f} K  (converged in {iters} iters)")
    print(f"  dof_tot_initial   = {dof_tot_init:.4f},  dof_tot_eq = {dof_tot_eq:.4f}")
    print()
```
